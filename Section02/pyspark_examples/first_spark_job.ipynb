```python
from pyspark.sql import SparkSession

spark = SparkSession.builder.appName("Read Inside Airbnb data").getOrCreate()
```

```python
listings = spark.read.csv("listings.csv.gz", header=True, inferSchema=True, sep=",", quote='"', escape='"', multiLine=True, mode="PERMISSIVE")
```

```python
review_locations = listings.select(listings.review_scores_location)
review_locations.show()
```

```python
high_score_listings = listings.filter(listings.review_scores_location > 4.5).select('id', 'price', 'name', 'review_scores_location')
high_score_listings.show(20, truncate=False)
```

```python
high_score_listings.dropna().show(20, truncate=False)
```

```python
from pyspark.sql.functions import regexp_replace

price_num_df = listings.withColumn('price_num', regexp_replace('price', '[$,]', '').cast('float'))
price_num_df.schema['price_num']
price_num_df.select('price_num', 'name').show(20, truncate=False)
```

```python
price_num_df.filter((price_num_df.price_num < 100) & (price_num_df.review_scores_location > 4.5)).select('name', 'price', 'review_scores_location').show(truncate=False)
```

```python
price_num_df.filter('price_num < 100 AND review_scores_location > 4.5').select('name', 'price', 'review_scores_location').show(truncate=False)
```

```python
listings.select(listings.property_type).distinct().show(truncate=False)
```

```python
listings.select(listings.property_type, listings.room_type).distinct().show(truncate=False)
```

```python
listings.select(listings.property_type).distinct().write.csv('property_types')
```
